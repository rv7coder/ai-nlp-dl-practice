```python
import spacy
```

```python
import pandas as pd
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
para = "Google DeepMind is placing greater emphasis on the ethical development of artificial intelligence by integrating philosophers and ethics researchers into its AI teams. One of the key figures is philosopher Iason Gabriel, who works on ensuring that advanced AI systems are not only powerful but also aligned with human values. His work focuses on issues such as fairness, accountability, transparency, and the social impact of AI. As AI models become more capable, experts argue that technical improvements alone are not enough. Developers must also consider how AI affects society, employment, privacy, and decision-making. Google DeepMind believes that combining philosophy with computer science can help create AI systems that are safer, more reliable, and beneficial for everyone. This reflects a growing trend across the AI industry to prioritize responsible AI development alongside rapid technological innovation."
```

```python
para
```

```python
docs = nlp(para)
```

```python
df_docs = pd.DataFrame(columns = ('tokens', 'pos_tag'))
```

```python
df_docs.head()
```

```python
for token in docs:
    df_docs = pd.concat([df_docs, pd.DataFrame.from_records([{'tokens': token.text, 'pos_tag': token.pos_}])], ignore_index=True)
```

```python
df_docs.head(20)
```

```python
df_docs_counts = df_docs.groupby(['tokens', 'pos_tag']).size().reset_index(name = "counts").sort_values(by = "counts", ascending = False)
df_docs_counts.head(20)
```

```python
nouns = df_docs_counts[df_docs_counts.pos_tag == "NOUN"][0:25]
```
