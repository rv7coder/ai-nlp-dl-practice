```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
import re
import spacy
import pandas as pd
```

```python
para = "Bengaluru is now the leading hub for India's youngest founders, with 21 entrepreneurs from the city featuring on the Avendus Wealth - Hurun India U30 List 2026, up from just 7 last year, Priyank Kharge, Minister of Home, e-Governance, Electronics, IT & BT, Government of Karnataka, said.Kharge cited the report to highlight a broader shift in India's innovation landscape. According to the Avendus Wealth - Hurun India U30 List 2026, which recognises 102 exceptional leaders aged 30 and under, 18 of the companies on the list are headquartered in Bengaluru -- the highest for any city in the country. The list brings together first-generation founders building ventures valued at USD 25 million or more and next-generation leaders driving family-owned businesses valued at USD 50 million or more.The list reflects how Bengaluru continues to lead from the front, Kharge said. This is a testament to Bengaluru's entrepreneurial spirit, world-class talent and collaborative innovation ecosystem.The report points to a clear pivot toward DeepTech and HardTech. One in four entrants on this year's list of 27 founders, came from DeepTech and HardTech sectors, including AI & ML, EV & Auto Components, SpaceTech, Aerospace & Defence and Cybersecurity. The number of entrepreneurs from AI and Machine Learning ventures has nearly doubled this year to 8, while SpaceTech features 6 founders across Pixxel, Digantara and Apolink. The Automobile & Auto Components sector also doubled to 7 entrants, with all new entrants focused on electric vehicles or components."
print(para)
```

```python
para.lower()
```

```python
data = pd.DataFrame({'text': [para]})
data.head()
```

```python
data['text_lowercase'] = data['text'].str.lower()
data.head()
```

```python
data['text_no_punct'] = data.apply(lambda x: re.sub(r'[^\w\s]', '', x['text_lowercase']), axis = 1)
data.head()
```

```python
en_stopwords = stopwords.words('english')
```

```python
data['text_no_punct_no_stopwords'] = data['text_no_punct'].apply(lambda x: ' '.join([word for word in x.split() if word not in (en_stopwords)]))
data['text_no_punct_no_stopwords'][0]
```

```python
data['text_tokens'] = data.apply(lambda x: word_tokenize(x['text_no_punct_no_stopwords']), axis = 1)
data['text_tokens'][0]
```

```python
lemmatizer = WordNetLemmatizer()
data['tokens_lemmatized'] = data['text_tokens'].apply(lambda tokens: [lemmatizer.lemmatize(token) for token in tokens])
data['tokens_lemmatized'][0]
```

```python
new_df = pd.DataFrame(columns=['tokens_cl', 'pos_tags'])
new_df.head()
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
data['final_tokens'] = data['tokens_lemmatized'].apply(lambda x: ' '.join(x))
data.head()
```

```python
final_tokens_list = sum(data['tokens_lemmatized'], [])
print(final_tokens_list)
```

```python
spacy_docs_tokens = nlp(' '.join(final_tokens_list))
```

```python
for word in spacy_docs_tokens:
    new_df = pd.concat([new_df, pd.DataFrame.from_records([{'tokens_cl': word.text, 'pos_tags': word.pos_}])], ignore_index=True)
new_df.head(20)    
```

```python
new_df_counts = new_df.groupby(['tokens_cl', 'pos_tags']).size().reset_index(name='counts').sort_values(by='counts', ascending = False)
new_df_counts.head(20)
```

```python
new_df_ner = pd.DataFrame(columns=['tokens_cl', 'ner_tags'])
```

```python
for word in spacy_docs_tokens.ents:
    if pd.isna(word.label_) is False:
        new_df_ner = pd.concat([new_df_ner, pd.DataFrame.from_records([{'tokens_cl': word.text, 'ner_tags': word.label_}])], ignore_index = True)
new_df_ner.head(20)    
```
